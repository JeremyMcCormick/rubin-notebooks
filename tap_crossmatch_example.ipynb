```python
Irsa.query_tap(query="SELECT schema_name, table_name FROM TAP_SCHEMA.tables WHERE tables.schema_name LIKE 'ztf%'")
```

```python
ztf_radec = Irsa.query_tap(query="SELECT column_name, ucd FROM TAP_SCHEMA.columns WHERE columns.table_name = 'ztf_objects_dr23'"
                           " AND (ucd LIKE 'pos.eq.ra' OR ucd LIKE 'pos.eq.dec') ORDER BY ucd DESC").to_qtable().to_pandas()
```

```python
ztf_ra_colname = ztf_radec[0][0]
ztf_dec_colname = ztf_radec[1][0]

print(f"ZTF ra is '{ztf_ra_colname}' and dec is '{ztf_dec_colname}'")
```

```python
def get_radec_colnames(table_name):
    query = f"SELECT column_name, ucd FROM TAP_SCHEMA.columns WHERE columns.table_name = '{table_name}' AND (ucd LIKE 'pos.eq.ra' OR ucd LIKE 'pos.eq.dec') ORDER BY ucd DESC"
    radec = Irsa.query_tap(query=query).to_qtable().to_pandas()
    return radec.iloc[0, 0], radec.iloc[1, 0]
```

```python
ztf_ra_colname, ztf_dec_colname = get_radec_colnames('ztf_objects_dr23')
print(f"ZTF ra is '{ztf_ra_colname}' and dec is '{ztf_dec_colname}'")
```
