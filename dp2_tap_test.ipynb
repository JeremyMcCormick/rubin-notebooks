```python
from lsst.rsp import get_tap_service
```

```python
service = get_tap_service("tap")
```

```python
job = service.submit_job("SELECT table_name FROM tap_schema.tables WHERE schema_name LIKE '%dp2'")
job.run()
job.wait(phases=['COMPLETED', 'ERROR'])
if job.phase == "ERROR":
    job.raise_if_error() 
tables_result = job.fetch_result().to_table()
```

```python
tables_result
```

```python
for table_name in tables_result["table_name"]:
    print(f"Checking {table_name}...")
    job = service.submit_job(f"SELECT TOP 1 * FROM {table_name}")
    job.run()
    job.wait(phases=['COMPLETED', 'ERROR'])
    try:
        if job.phase == "ERROR":
            job.raise_if_error()
        table = job.fetch_result().to_table()
        print(table)
    except Exception as e:
        print(e)
    print(f"Done checking {table_name}\n")
```
