```python
from lsst.rsp import get_tap_service
from IPython.display import display, HTML
```

```python
service = get_tap_service("tap")
```

```python
schema_name = "dp1"
schema_query = f"SELECT * FROM tap_schema.schemas WHERE schema_name = '{schema_name}'"
schema_results_table = service.search(schema_query).to_table()

table_query = f"SELECT * FROM tap_schema.tables WHERE tap_schema.tables.schema_name = '{schema_name}' ORDER BY table_index ASC"
table_results_table = service.search(table_query).to_table()
for table_name in table_results_table["table_name"]:
    test_query = f"SELECT COUNT(*) FROM {table_name}"
    try:
        test_results_table = service.search(test_query).to_table()
        print(f"{table_name}: {test_results_table}")
    except Exception as e:
         print("Error selecting from %s: %s" % (table_name, e))
    print("-"*40)
```

```python
column_query = f"SELECT DISTINCT(table_name) FROM tap_schema.columns GROUP BY table_name"
column_results_table = service.search(column_query).to_table()
```

```python
dp1_column_query = """SELECT
  table_name,
  COUNT(column_name) AS column_count
FROM tap_schema.columns
WHERE table_name like 'dp1.%'
GROUP BY table_name
ORDER BY column_count DESC;
"""
dp1_column_results_table = service.search(dp1_column_query).to_table()
dp1_column_results_table
```
